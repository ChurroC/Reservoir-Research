```python
import numpy as np
from scipy.optimize import root
import plotly.graph_objects as go
```

```python
def general_system_equations(X, epsilons, kappa, kappa_wall):
    """Defines steady-state force balance for an N-node chain."""
    N = len(X)
    eqs = np.zeros(N)

    for i in range(N):
        Xi = X[i]
        eps_i = epsilons[i]

        # 1. Nonlinear von Mises truss force for node i
        truss_force = -2 * (1 - (1 + eps_i) / np.sqrt(1 + Xi**2)) * Xi

        # Left and wall
        if i == 0:
            left_force = kappa * Xi + kappa_wall * Xi
        else:
            left_force = kappa * (Xi - X[i - 1])

        # Right and wall
        if i == N - 1:
            right_force = kappa * Xi + kappa_wall * Xi  # Anchored to right wall
        else:
            right_force = kappa * (Xi - X[i + 1])

        eqs[i] = truss_force - left_force - right_force

    return eqs


# --- Parameters for 10 Nodes ---
N = 10
epsilons = np.ones(N) * .1
kappa = 1.0
kappa_wall = 0.0

rng = np.random.default_rng(seed=42)
guess_count = 10000
guesses = rng.uniform(-1.0, 1.0, (guess_count, N))

unique_solutions = []

print(f"{'Guess #':<8} | {'Solver Status':<25} | {'Unique Steady State Found?'}")
print("-" * 60)

for idx, guess in enumerate(guesses):
    sol = root(general_system_equations, guess, args=(epsilons, kappa, kappa_wall))

    rounded_sol = np.round(sol.x, 4)

    is_new = True
    for u in unique_solutions:
        if np.allclose(u, rounded_sol, atol=1e-3):
            is_new = False
            break

    if is_new and sol.success:
        unique_solutions.append(rounded_sol)

    if idx%(guess_count//10)==0:
        print(
            f"{idx+1:<8} | {str(sol.message):<25} | {str(is_new)}" f" (Converged to state)"
        )

print("\n--- Unique Steady States Discovered ---")
for idx, state in enumerate(unique_solutions):
    print(f"State {idx+1}: {state}")
```

```python
def check_controllability_10_nodes(X_steady, epsilons, kappa, zeta, B_matrix):
    N = len(X_steady)

    # 1. Compute local stiffness derivative for each node
    k_local = np.zeros(N)
    for i in range(N):
        Xi = X_steady[i]
        eps_i = epsilons[i]
        k_local[i] = -2 * (1 - (1 + eps_i) / (1 + Xi**2) ** 1.5)

    # 2. Build the N x N effective stiffness matrix K_eff
    K_eff = np.zeros((N, N))
    for i in range(N):
        deg = 2 if (0 < i < N - 1) else 1
        K_eff[i, i] = k_local[i] - deg * kappa
        if i > 0:
            K_eff[i, i - 1] = kappa
        if i < N - 1:
            K_eff[i, i + 1] = kappa

    # 3. Build the full 2N x 2N system matrix A
    A = np.block([[np.zeros((N, N)), np.eye(N)], [K_eff, -2 * zeta * np.eye(N)]])

    # 4. Construct the Kalman Controllability Matrix C
    n_states = A.shape[0]
    C = B_matrix
    current_term = B_matrix
    for i in range(1, n_states):
        current_term = A @ current_term
        C = np.hstack((C, current_term))

    # 5. Compute Matrix Rank
    rank = np.linalg.matrix_rank(C)
    return rank


# --- Parameters ---
N = 10
epsilons = np.full(N, 0.1)
zeta = 0.5

# Actuator on Node 1 velocity (V1)
B_matrix = np.zeros((2 * N, 1))
B_matrix[N, 0] = 1.0

# Define Steady States
states = {
    "All Center Peak": np.zeros(N),
    "All Down Well": np.full(N, -0.4583),
    "All Up Well": np.full(N, 0.4583),
}

# --- Logarithmic Sweep for Kappa ---
kappa_values = np.logspace(0, -5, 50)
rank_results = {name: [] for name in states}

for kappa in kappa_values:
    for name, state in states.items():
        rank = check_controllability_10_nodes(state, epsilons, kappa, zeta, B_matrix)
        rank_results[name].append(rank)

# --- Plotly Interactive Figure ---
fig = go.Figure()

for name, ranks in rank_results.items():
    fig.add_trace(
        go.Scatter(
            x=kappa_values,
            y=ranks,
            mode="lines+markers",
            name=name,
            marker=dict(size=6),
            line=dict(width=2),
        )
    )

# Layout configuration
fig.update_layout(
    title=("<b>Controllability Rank Collapse vs. Coupling Stiffness" " (&kappa;)</b>"),
    xaxis_title="Coupling Stiffness k [Log Scale]",
    yaxis_title="Controllability Matrix Rank (Max 20)",
    xaxis_type="log",
    template="plotly_white",
    hovermode="x unified",
    yaxis=dict(range=[0, 22]),
)

# Reverse x-axis so strong coupling sits on the left
fig.update_xaxes(autorange="reversed")

# Render the interactive plot in your browser/notebook
fig.show()
```
