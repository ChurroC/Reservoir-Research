```python
import os
from pathlib import Path


def generate_experiment_markdown(exp_dir_path):
    exp_path = Path(exp_dir_path)
    exp_name = exp_path.name

    print(f"- ## Files")

    # Find all Jupyter notebooks or python files in the experiment folder
    code_files = sorted(
        [f for f in exp_path.iterdir() if f.suffix in [".ipynb", ".py"]]
    )

    for code in code_files:
        index = code.name[0]
        print(f"  - `{code.name}`")

        print(f"    - blah blah")

        plots_path = exp_path / "plots"
        if plots_path.exists():
            plots = sorted(
                [
                    p.name
                    for p in plots_path.iterdir()
                    if p.suffix in [".png", ".jpg", ".jpeg"]
                    and p.name.startswith(index)
                ]
            )

            for plot in plots:
                print(f"    - `{plot}`")
                print(f"      - blah blah")
                print(f"      - ![{plot}]({exp_dir_path}/plots/{plot}){{:width 450}}")
```
