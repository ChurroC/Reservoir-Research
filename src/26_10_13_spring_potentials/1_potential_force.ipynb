```python
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots
```

```python
def plot_potential():
    # 3. Create the plot figure
    fig = go.Figure()

    # Add the continuous curve for the potential well
    fig.add_trace(
        go.Scatter(
            x=x_vals,
            y=y_vals,
            mode="lines",
            name="Potential Energy V(x)",
            line=dict(color="#3b82f6", width=3),
        )
    )

    # Highlight the Stable Minima (The Wells at x = -1 and x = 1)
    fig.add_trace(
        go.Scatter(
            x=[-1, 1],
            y=[potential_func(-1), potential_func(1)],
            mode="markers",
            name="Stable Minima (Wells)",
            marker=dict(color="#10b981", size=12, symbol="circle"),
        )
    )

    # Highlight the Unstable Maximum (The Energy Barrier at x = 0)
    fig.add_trace(
        go.Scatter(
            x=[0],
            y=[potential_func(0)],
            mode="markers",
            name="Unstable Maximum (Barrier)",
            marker=dict(color="#ef4444", size=12, symbol="x"),
        )
    )

    # 4. Style the layout, axes, and grid lines
    fig.update_layout(
        title=dict(
            text="Bistable Potential Well: V(x) = x⁴ - 2x²", x=0.5, font=dict(size=18)
        ),
        xaxis=dict(
            title="Position (x)",
            gridcolor="rgba(200, 200, 200, 0.2)",
            zerolinecolor="rgba(100, 100, 100, 0.3)",
        ),
        yaxis=dict(
            title="Potential Energy V(x)",
            gridcolor="rgba(200, 200, 200, 0.2)",
            zerolinecolor="rgba(100, 100, 100, 0.3)",
        ),
        template="plotly_white",
        hovermode="x unified",
        legend=dict(yanchor="top", y=0.99, xanchor="left", x=0.01),
    )

    # 5. Render the chart
    fig.show()
```

```python
def plot_bistable_and_force():
    # 4. Create subplots (1 row, 2 columns)
    fig = make_subplots(
        rows=1,
        cols=2,
        subplot_titles=("Potential Well V(x)", "Derivative dV/dx (Force Profile)"),
    )

    # Left Subplot: The Bistable Well
    fig.add_trace(
        go.Scatter(
            x=x_vals,
            y=y_vals,
            name="Potential Energy V(x)",
            line=dict(color="#3b82f6", width=3),
        ),
        row=1,
        col=1,
    )

    # Right Subplot: The Exact Analytical Derivative
    fig.add_trace(
        go.Scatter(
            x=x_vals,
            y=dv_dx_analytical,
            name="Analytical dV/dx",
            line=dict(color="#10b981", width=3),
        ),
        row=1,
        col=2,
    )

    # Right Subplot: The Numerical np.gradient (dashed line to check overlay)
    fig.add_trace(
        go.Scatter(
            x=x_vals,
            y=dv_dx_numerical,
            name="np.gradient()",
            line=dict(color="#ef4444", width=2, dash="dash"),
        ),
        row=1,
        col=2,
    )

    # 5. Styling Layout
    fig.update_layout(
        title=dict(
            text="Bistable Potential Well vs. Its Derivative", x=0.5, font=dict(size=18)
        ),
        template="plotly_white",
        hovermode="x unified",
    )

    # Update axes titles
    fig.update_xaxes(title_text="Position (x)", row=1, col=1)
    fig.update_yaxes(title_text="V(x)", row=1, col=1)
    fig.update_xaxes(title_text="Position (x)", row=1, col=2)
    fig.update_yaxes(title_text="dV/dx", row=1, col=2)

    fig.show()
```

```python
# 1. Define the bistable potential function: V(x) = x^4 - 2x^2
def potential_func(x):
    return x**4 - 2 * x**2


# 2. Generate data points for the curve
x_vals = np.linspace(-1.5, 1.5, 400)
y_vals = potential_func(x_vals)

dv_dx_numerical = -np.gradient(y_vals, x_vals)
dv_dx_analytical = -(4 * x_vals**3 - 4 * x_vals)
```

```python
plot_potential()
plot_bistable_and_force()
```

```python
k = 10.0  # Stiffness
h = 1.0  # Fixed anchor height
L0 = 1.5  # Uncompressed rest length (must be > h to form a bistable well)

def potential_func(x):
    return k * (np.sqrt(x**2 + h**2) - L0) ** 2


# 2. Generate data points for the curve
x_vals = np.linspace(-1.5, 1.5, 400)
y_vals = potential_func(x_vals)

dv_dx_numerical = -np.gradient(y_vals, x_vals)
dv_dx_analytical = -(2 * k * x_vals * (1 - L0 / np.sqrt(x_vals**2 + h**2)))
```

```python
plot_potential()
plot_bistable_and_force()
```
