```python
import numpy as np

from numba import njit

from sklearn.linear_model import RidgeCV
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_squared_error, accuracy_score

import plotly.graph_objects as go

import fastplotlib as fpl
```

# Initial Idea

```python
time = 1000
dt = 0.005
steps = int(time / dt)
tau_steps = int(1.5 / dt)

test_size = 0.2

node_grid_len = 3 # Then a 2 by 2 grid
grid_dim = 2
total_node = node_grid_len ** grid_dim
matrix_size = (total_node) * grid_dim

m_diag = np.random.uniform(0.1, 0.5, size=matrix_size)
m_inv_diag = 1.0 / m_diag
M_INV = np.diag(m_inv_diag)

DAMP = np.diag(np.random.uniform(0.1, 0.3, size=matrix_size))

t = np.linspace(0, time, steps)

u_val = 1
u = (t.astype(int) % 2) * 2 * u_val - u_val
```

```python
K = np.zeros((matrix_size, matrix_size))
k_vals = np.random.uniform(0.5, 100, 2 * (node_grid_len - 1) * (2 * node_grid_len - 1))
k_val_index = 0

for i in range(1, total_node + 1):
    # check left
    j = i - 1 # For example node_i = 2 node_j = 1
    if 1 <= j % node_grid_len < node_grid_len:
        # Now for spring 2 to 1
        # pos is for matrix position not nodal pos
        i_pos = i * 2 - 2 # for spring 2 we get index 2
        j_pos = j * 2 - 2  # for spring 1 we get index 0
        K[i_pos, i_pos] += k_vals[k_val_index]
        K[i_pos, j_pos] += -k_vals[k_val_index]
        K[j_pos, j_pos] += k_vals[k_val_index]
        K[j_pos, i_pos] += -k_vals[k_val_index]
        k_val_index += 1

    # check top
    j = i - node_grid_len
    if j > 0 and 0 <= j % node_grid_len < node_grid_len:
        # print(f"i: {i}, j: {j}")
        i_pos = i * 2 - 2 + 1
        j_pos = j * 2 - 2 + 1
        K[i_pos, i_pos] += k_vals[k_val_index]
        K[i_pos, j_pos] += -k_vals[k_val_index]
        K[j_pos, j_pos] += k_vals[k_val_index]
        K[j_pos, i_pos] += -k_vals[k_val_index]
        k_val_index += 1

    j = i - node_grid_len - 1
    if j > 0 and (i - 1) % node_grid_len > 0:
        k_shear = k_vals[k_val_index]
        k_val_index += 1

        ix = i * 2 - 2
        iy = i * 2 - 1
        jx = j * 2 - 2
        jy = j * 2 - 1

        k_comp = 0.5 * k_shear

        K[ix, ix] += k_comp
        K[iy, iy] += k_comp
        K[ix, iy] += k_comp
        K[iy, ix] += k_comp 

        K[jx, jx] += k_comp
        K[jy, jy] += k_comp
        K[jx, jy] += k_comp
        K[jy, jx] += k_comp

        K[ix, jx] -= k_comp
        K[ix, jy] -= k_comp
        K[iy, jx] -= k_comp
        K[iy, jy] -= k_comp
        K[jx, ix] -= k_comp
        K[jx, iy] -= k_comp
        K[jy, ix] -= k_comp
        K[jy, iy] -= k_comp

    j = i - node_grid_len + 1
    if j > 0 and (i % node_grid_len) > 0:
        k_shear = k_vals[k_val_index]
        k_val_index += 1

        ix = i * 2 - 2
        iy = i * 2 - 1
        jx = j * 2 - 2
        jy = j * 2 - 1
        k_comp = 0.5 * k_shear

        K[ix, ix] += k_comp
        K[iy, iy] += k_comp
        K[ix, iy] -= k_comp
        K[iy, ix] -= k_comp

        K[jx, jx] += k_comp
        K[jy, jy] += k_comp
        K[jx, jy] -= k_comp
        K[jy, jx] -= k_comp

        K[ix, jx] -= k_comp
        K[ix, jy] += k_comp
        K[iy, jx] += k_comp
        K[iy, jy] -= k_comp
        K[jx, ix] -= k_comp
        K[jx, iy] += k_comp
        K[jy, ix] += k_comp
        K[jy, iy] -= k_comp
```

```python
@njit
def run_simulation(steps, dt, u, M_INV, K, C, N):
    x = np.zeros((steps, N))
    v = np.zeros((steps, N))
    acc = np.zeros(N)
    u_force = np.zeros(N)
    u_force_next = np.zeros(N)

    for i in range(1, steps):
        u_force[0] = u[i - 1]
        u_force[1] = u[i - 1]

        acc = M_INV @ (-K @ x[i - 1] - C @ v[i - 1] + u_force)

        x[i] = x[i - 1] + v[i - 1] * dt + acc * 0.5 * dt**2

        u_force_next[0] = u[i]
        u_force_next[1] = u[i]

        acc_next = M_INV @ (-K @ x[i] - C @ (v[i - 1] + acc * dt) + u_force_next)

        v[i] = v[i - 1] + 0.5 * (acc + acc_next) * dt

    return x, v
```

```python
x, v = run_simulation(steps, dt, u, M_INV, K, DAMP, matrix_size)
```

```python
xx, yy = np.meshgrid(np.arange(node_grid_len), np.arange(node_grid_len))
x_init = xx.flatten()
y_init = yy.flatten()
dots_data = np.column_stack([x_init, y_init]).astype(np.float32)

hor = x[0::2]
ver = x[1::2]

fig = fpl.Figure()
dots_graphic = fig[0, 0].add_scatter(data=dots_data, sizes=15, colors="magenta")

frame_tracker = 0
frames = 5
max_frames = 2000

def update_springs(canvas):
    global frame_tracker

    frame_tracker = (frame_tracker + frames) % steps
    print(f"Frame: {frame_tracker}/{steps}", end="\r")

    if frame_tracker >= max_frames:
        canvas.clear_animations()

    current_state = x[frame_tracker]
    hor = current_state[0::2]
    ver = current_state[1::2]

    dots_graphic.data[:, 0] = x_init + hor
    dots_graphic.data[:, 1] = y_init + ver

fig.add_animations(update_springs)
fig.show()
```

# Modifiable K

```python
time = 1000
dt = 0.005
steps = int(time / dt)
tau_steps = int(1.5 / dt)

test_size = 0.2

t = np.linspace(0, time, steps)

u_val = 1
u = (t.astype(int) % 2) * 2 * u_val - u_val
```

```python
"""
Example nodes
sample_nodes = np.array(
    [
        [0.0, 1.0],  # Node 0: Top-Left
        [1.0, 1.0],  # Node 1: Top-Right
        [0.0, 0.0],  # Node 2: Bottom-Left
        [1.0, 0.0],  # Node 3: Bottom-Right
    ]
)

sample_connections = np.array(
    [
        (0, 1, 93.7),  # Top horizontal spring
        (2, 3, 34.4),  # Bottom horizontal spring
        (0, 2, 14.2),  # Left vertical spring
        (1, 3, 11.8),  # Right vertical spring
        (0, 3, 25.0),  # Diagonal shear 1
        (1, 2, 25.0),  # Diagonal shear 2
    ]
)
"""
def create_stiffness_matrix(node_positions, connections):
    num_nodes = node_positions.shape[0]
    dims = node_positions.shape[1]
    K = np.zeros((num_nodes * dims, num_nodes * dims))

    for conn in connections:
        node_conn = conn[:2].astype(int)
        node_pos = node_positions[node_conn]
        k_val = conn[2]

        # We do this direction instead of the sin and cos in usual 2D
        # Diff is just subtracting next point by current point or a vec point to next vec
        diff_vec = np.diff(node_pos, axis=0).flatten()
        unit_dir = diff_vec / np.linalg.norm(diff_vec)

        """
        cc, ss, cs = c * c, s * s, c * s
            k_local = k_val * np.array([
                [ cc,  cs, -cc, -cs],
                [ cs,  ss, -cs, -ss],
                [-cc, -cs,  cc,  cs],
                [-cs, -ss,  cs,  ss]
            ])
        This is the matrix for a 2D spring with rotation
        So instead of the k_hor and k_ver we can use this with any angle
        But also if you look at it doing an outer product is the same as this
        So we keep [c, s] then do a outer product for the 2 by 2 sub grid
        """
        sub_block = np.outer(unit_dir, unit_dir)
        k_local = k_val * np.block([[sub_block, -sub_block], [-sub_block, sub_block]])

        """
        ix, iy = node_i * 2, node_i * 2 + 1
        jx, jy = node_j * 2, node_j * 2 + 1
        global_indices = [ix, iy, jx, jy]
        Basially for example between node 1 and 3 the ver spring
        It says the indices are [0, 1, 4, 5]
        And we would have the k vals in 1 and 5 since hor but depends on angle for all
        """
        # Basically same thing from before we take a pair then broadcast to dimension and add 1 as we go
        # Then we flatting col by col using "F"
        global_indices = (node_conn * dims + np.arange(dims)[:, None]).flatten("F")

        # This is just for every indices but the local just goes from 1 to dim for the local_grid
        for local_row, global_row in enumerate(global_indices):
            for local_col, global_col in enumerate(global_indices):
                K[global_row, global_col] += k_local[local_row, local_col]

    return K
```

```python
x = np.arange(10)
y = np.arange(10)

xx, yy = np.meshgrid(x, y)

nodes_pos = np.column_stack((xx.flatten(), yy.flatten()))
# nodes_pos = x.reshape(10, 1)
```

```python
num_nodes = nodes_pos.shape[0]
dims = nodes_pos.shape[1]
matrix_size = num_nodes * dims

m_diag = np.random.uniform(0.1, 0.5, size=matrix_size)
m_inv_diag = 1.0 / m_diag
M_INV = np.diag(m_inv_diag)

DAMP = np.diag(np.random.uniform(0.1, 0.3, size=matrix_size))
```

```python
node_ids = np.arange(x.size * y.size).reshape(
    y.size, x.size
)

src_nodes = np.concatenate(
    [
        node_ids[:, :-1].flatten(),  # Right links source
        node_ids[:-1, :].flatten(),  # Down links source
        node_ids[:-1, :-1].flatten(),  # Down-Right diagonals source
        node_ids[:-1, 1:].flatten(),  # Down-Left diagonals source
    ]
)

dst_nodes = np.concatenate(
    [
        node_ids[:, 1:].flatten(),  # Right links destination
        node_ids[1:, :].flatten(),  # Down links destination
        node_ids[1:, 1:].flatten(),  # Down-Right diagonals destination
        node_ids[1:, :-1].flatten(),  # Down-Left diagonals destination
    ]
)

k_vals = np.random.uniform(0.5, 8, size=src_nodes.shape[0])

connections_list = np.column_stack((src_nodes, dst_nodes, k_vals))
```

```python
K = create_stiffness_matrix(nodes_pos, connections_list)
```

```python
@njit
def run_simulation(steps, dt, u, M_INV, K, C, N):
    x = np.zeros((steps, N))
    v = np.zeros((steps, N))
    acc = np.zeros(N)
    u_force = np.zeros(N)
    u_force_next = np.zeros(N)

    for i in range(1, steps):
        u_force[0] = u[i - 1]
        # u_force[1] = u[i - 1]

        acc = M_INV @ (-K @ x[i - 1] - C @ v[i - 1] + u_force)

        x[i] = x[i - 1] + v[i - 1] * dt + acc * 0.5 * dt**2

        u_force_next[0] = u[i]
        # u_force_next[1] = u[i]

        acc_next = M_INV @ (-K @ x[i] - C @ (v[i - 1] + acc * dt) + u_force_next)

        v[i] = v[i - 1] + 0.5 * (acc + acc_next) * dt

    return x, v
```

```python
disp, v = run_simulation(steps, dt, u, M_INV, K, DAMP, matrix_size)
```

```python
disp_reshaped = disp.reshape(steps, num_nodes, dims)
disp_3d = np.pad(disp_reshaped, ((0, 0), (0, 0), (0, 3 - dims)), mode="constant")

nodes_pos_3D = np.pad(nodes_pos, ((0, 0), (0, 3 - dims)), mode="constant")

fig = fpl.Figure()
dots_graphic = fig[0, 0].add_scatter(data=nodes_pos_3D.astype(np.float32), sizes=15, colors="magenta")

spring_lines = []
src_indices = connections_list[:, 0].astype(np.int32)
dst_indices = connections_list[:, 1].astype(np.int32)
for src, dst in zip(src_indices, dst_indices):
    line_coords = np.vstack([nodes_pos_3D[src], nodes_pos_3D[dst]])
    line_graphic = fig[0, 0].add_line(data=line_coords.astype(np.float32), thickness=2, colors="cyan")
    line_graphic.colors[:, 3] = .2
    spring_lines.append(line_graphic)


frame_tracker = 0
frames_moved = 5
max_frames = 2000
def update_springs(canvas):
    global frame_tracker

    frame_tracker = (frame_tracker + frames_moved) % steps
    # print(f"Frame: {frame_tracker}/{steps}", end="\r")

    if frame_tracker >= max_frames:
        canvas.clear_animations()

    new_coords = nodes_pos_3D + disp_3d[frame_tracker]

    dots_graphic.data = new_coords.astype(np.float32)

    for idx, (src, dst) in enumerate(zip(src_indices, dst_indices)):
        src_coord = new_coords[src]
        dst_coord = new_coords[dst]
        spring_lines[idx].data = np.vstack([src_coord, dst_coord]).astype(np.float32)


fig.add_animations(update_springs)
fig.show()
```

```python
X = np.column_stack((disp, v))

scaler = StandardScaler()

X_train_unscaled, X_test_unscaled, y_train, y_test = train_test_split(
    X[tau_steps:], u[:-tau_steps], test_size=test_size, random_state=42
)

X_train = scaler.fit_transform(X_train_unscaled)
X_test = scaler.transform(X_test_unscaled)
```

```python
ridge_model = RidgeCV()
ridge_model.fit(X_train, y_train)
y_pred_ridge = ridge_model.predict(X_test)
```

```python
y_pred_discrete = np.where(y_pred_ridge > 0.0, u_val, -u_val)

r_2 = r2_score(y_test, y_pred_ridge)
accuracy = accuracy_score(y_test, y_pred_discrete) * 100

print(f"{r_2:.4f}", f"{accuracy:.2f}%")
```

```python
weights = ridge_model.coef_
labels = [
    f"{'Pos' if i % 2 == 0 else 'Vel'} Node {i//2 + 1}" for i in range(len(weights))
]

fig = go.Figure(
    data=[
        go.Bar(
            x=labels,
            y=weights,
            marker_color=np.where(
                weights >= 0, "royalblue", "firebrick"
            ),  # Blue for +, Red for -
        )
    ]
)

fig.update_layout(
    title="Reservoir Node Contribution (Feature Weights)",
    xaxis_title="Spring/Mass Node",
    yaxis_title="Weight Value",
    template="plotly_white",
)

fig.show()
```

# 3D Cube

```python
x = np.arange(3)
y = np.arange(3)
z = np.arange(3)

xx, yy, zz = np.meshgrid(x, y, z)

nodes_pos = np.column_stack((xx.flatten(), yy.flatten(), zz.flatten()))
```

```python
np.random.seed(42)

num_nodes = nodes_pos.shape[0]
dims = nodes_pos.shape[1]
matrix_size = num_nodes * dims

m_diag = np.random.uniform(0.1, 0.5, size=matrix_size)
m_inv_diag = 1.0 / m_diag
M_INV = np.diag(m_inv_diag)

DAMP = np.diag(np.random.uniform(0.1, 0.3, size=matrix_size))

k_vals = np.random.uniform(0.5, 8, size=src_nodes.shape[0])
```

```python
node_ids = np.arange(z.size * y.size * x.size).reshape(z.size, y.size, x.size)

src_slices = [
    np.s_[:, :, :-1],  # Right source
    np.s_[:, :-1, :],  # Down source
    np.s_[:-1, :, :],  # Deep source
    np.s_[:, :-1, :-1],  # Down-Right source
    np.s_[:-1, :-1, :-1],  # Deep-Down-Right source
]

dst_slices = [
    np.s_[:, :, 1:],  # Right destination
    np.s_[:, 1:, :],  # Down destination
    np.s_[1:, :, :],  # Deep destination
    np.s_[:, 1:, 1:],  # Down-Right destination
    np.s_[1:, 1:, 1:],  # Deep-Down-Right destination
]

src_list = []
dst_list = []

for src_sl, dst_sl in zip(src_slices, dst_slices):
    src_list.append(node_ids[src_sl].flatten())
    dst_list.append(node_ids[dst_sl].flatten())

src_nodes = np.concatenate(src_list)
dst_nodes = np.concatenate(dst_list)

connections_list = np.column_stack((src_nodes, dst_nodes, k_vals))

K = create_stiffness_matrix(nodes_pos, connections_list)
```

```python
disp, v = run_simulation(steps, dt, u, M_INV, K, DAMP, matrix_size)
```

```python
disp_reshaped = disp.reshape(steps, num_nodes, dims)
disp_3d = np.pad(disp_reshaped, ((0, 0), (0, 0), (0, 3 - dims)), mode="constant")

nodes_pos_3D = np.pad(nodes_pos, ((0, 0), (0, 3 - dims)), mode="constant")

fig = fpl.Figure(cameras="3d", controller_types="orbit")
dots_graphic = fig[0, 0].add_scatter(
    data=nodes_pos_3D.astype(np.float32), sizes=15, colors="magenta"
)

spring_lines = []
src_indices = connections_list[:, 0].astype(np.int32)
dst_indices = connections_list[:, 1].astype(np.int32)
for src, dst in zip(src_indices, dst_indices):
    line_coords = np.vstack([nodes_pos_3D[src], nodes_pos_3D[dst]])
    line_graphic = fig[0, 0].add_line(
        data=line_coords.astype(np.float32), thickness=2, colors="cyan"
    )
    line_graphic.colors[:, 3] = 0.2
    spring_lines.append(line_graphic)


frame_tracker = 0
frames_moved = 5
max_frames = 2000
def update_springs(canvas):
    global frame_tracker

    frame_tracker = (frame_tracker + frames_moved) % steps
    print(f"Frame: {frame_tracker}/{steps}", end="\r")

    if frame_tracker >= max_frames:
        canvas.clear_animations()

    new_coords = nodes_pos_3D + disp_3d[frame_tracker]

    dots_graphic.data = new_coords.astype(np.float32)

    for idx, (src, dst) in enumerate(zip(src_indices, dst_indices)):
        src_coord = new_coords[src]
        dst_coord = new_coords[dst]
        spring_lines[idx].data = np.vstack([src_coord, dst_coord]).astype(np.float32)


fig.add_animations(update_springs)
fig.show()
```
