# 100,000 steps and N is 1000 and timeout is 150

```python
def print_memory_estimate(steps, num_nodes, dims):
    matrix_size = num_nodes * dims
    bytes_per_float = 8
    # disp, v, and U matrices
    total_bytes = (steps * matrix_size * bytes_per_float) * 3 
    total_gb = total_bytes / (1024**3)
    print(f"Memory required for ONE trial simulation: {total_gb:.2f} GB")
    print(f"Memory required if running 50 concurrent cores: {total_gb * 50:.2f} GB")

print_memory_estimate(steps=100_000, num_nodes=1000, dims=2)
```
